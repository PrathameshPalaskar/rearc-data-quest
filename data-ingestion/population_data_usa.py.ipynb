```python
import requests
import json
import os

HEADERS = {"User-Agent": "Prathamesh Palaskar contact: Prathamesh Palaskar"}

# NOTE:This is is the correct docs linkPOP_URL = (
    "https://api.datausa.io/tesseract/data.jsonrecords"
    "?cube=acs_yg_total_population_1&drilldowns=Nation,Year&measures=Population"
)

LANDING_DIR = "/Volumes/rearc_bls/bronze/raw_landing/population"

os.makedirs(LANDING_DIR, exist_ok=True)

resp = requests.get(POP_URL, headers=HEADERS)
resp.raise_for_status()

with open(os.path.join(LANDING_DIR, "population.json"), "w") as f:
    json.dump(resp.json(), f)

print(f"Saved population data to {LANDING_DIR}/population.json")
```
